### Messages in LangChain

Messages are the basic units of communication between a user, an AI model, and tools. LangChain provides different message types, such as:

- **HumanMessage** — represents the user's input.
- **AIMessage** — represents the model's response, including tool calls.
- **ToolMessage** — contains the result returned by a tool.
- **SystemMessage** — provides instructions or behavior guidelines to the model.

Messages maintain the conversation history and allow LangChain to pass information between different steps of an AI application.

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain.tools import tool

load_dotenv()

# Create model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model(
    "groq:llama-3.3-70b-versatile"
)
```

```python
from langchain.messages import HumanMessage, SystemMessage, AIMessage

messages = [
    SystemMessage("You are a poetry expert!"),
    AIMessage("Write a poem about Artificial Intelligence")
]

response = model.invoke(messages)
response.content
```

```python
System_msg = SystemMessage("You are a helpful Assistant")

messages = [
    System_msg,
    HumanMessage("How do I create Rest APi")
]

response = model.invoke(messages)
print(response.content)
```

```python
# Detailed info to the LLM thrpugh System Message
System_msg = SystemMessage("""
You are a senior python developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but thorough in your explanations.
""")

messages = [
    System_msg,
    HumanMessage("How do I create REST API")
]

response = model.invoke(messages)
print(response.content)
```

```python
# Create an  AI message manually (e.g for coversation history)
ai_msg = ("I would be happy to help you with that question!")

messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great what's 2 + 2?")
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import ToolMessage
from langchain.messages import AIMessage

# After a model takes a tool call 
# (Here we demonstrate manually creating for brevity)

ai_message = AIMessage(
    content=[],
    tool_calls = [{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
    }]
)

# Execute tool and create reslut message
weather_result = "Sunny, 72"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id = "call_123" # Must match the call ID
)

# Continue Conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,
    tool_message
]

response = model.invoke(messages)
```
