### Model Integration with OpenAI, GROQ, Google Gemini

```python
import os
from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
model =  ChatGoogleGenerativeAI(

    model = "gemini-3.5-flash",
    temperature = 0
)

model
```

```python
response = model.invoke("How are you?")
response.content
```

### Streaming and Batch

```python
model.invoke("Write a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write a 200 words paragraph on Artificial Intelligence"):
    print(chunk.content, end="|", flush=True)
```

```python
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "What is quantum computing?",
    "How do plane fly?"
],
    config = {"max_concurrency": 5}

)

for response in responses:
    print(response)
```
